# Separate Male & Female Data:-

```python
male_sales = df[df["Gender"] == "Male"]["Total_Sales"]
female_sales = df[df["Gender"] == "Female"]["Total_Sales"]
```

```python
print("Male observations:", len(male_sales))
print("Female observations:", len(female_sales))
```

# Hypothesis Test:-

```python
t_stat, p_value = stats.ttest_ind(
    male_sales,
    female_sales,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

# Result Interpretation:-

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
    print("There is a statistically significant difference in average sales between male and female customers.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is no statistically significant difference in average sales between male and female customers.")
```

# 95% Confidence Interval:-

```python
mean_difference = male_sales.mean() - female_sales.mean()

se = np.sqrt(
    male_sales.var(ddof=1) / len(male_sales)
    +
    female_sales.var(ddof=1) / len(female_sales)
)

df_welch = (
    (male_sales.var(ddof=1) / len(male_sales)
     +
     female_sales.var(ddof=1) / len(female_sales)) ** 2
    /
    (
        (male_sales.var(ddof=1) / len(male_sales)) ** 2 / (len(male_sales) - 1)
        +
        (female_sales.var(ddof=1) / len(female_sales)) ** 2 / (len(female_sales) - 1)
    )
)

t_critical = stats.t.ppf(0.975, df_welch)

lower = mean_difference - t_critical * se
upper = mean_difference + t_critical * se

print("Mean Difference:", mean_difference)
print("95% Confidence Interval:", (lower, upper))
```

# Final Statistical Summary:-

```python
statistical_summary = pd.DataFrame({
    "Metric": [
        "Male Average Sales",
        "Female Average Sales",
        "Mean Difference",
        "T-statistic",
        "P-value",
        "Confidence Level"
    ],
    "Value": [
        male_sales.mean(),
        female_sales.mean(),
        mean_difference,
        t_stat,
        p_value,
        "95%"
    ]
})

statistical_summary
```
